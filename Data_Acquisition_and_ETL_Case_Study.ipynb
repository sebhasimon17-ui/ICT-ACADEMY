#Task 8

Merge all four datasets into one DataFrame.

Hint:

● Use the common column (country)

● Choose an appropriate join method

```python
merged_df = population.merge(internet_users, on="country", how="inner")\
.merge(GDP, on="country", how="inner")\
.merge(data_set, on="country", how="inner")
```

```python
merged_df
```

#Task 9

Create the following calculated columns:

1. Internet Penetration (%)

2. GDP Per Capita

Hint: Round the calculated values to two decimal places

```python
merged_df["internet_penetration (%)"] = (
    merged_df["internet_users"] / merged_df["population"] * 100
  ).round(2)
merged_df

```

```python
merged_df['gdp_per_capita'] = (
    merged_df['GDP'] / merged_df['population']
).round(2)
merged_df
```

#16 Find the country with the highest GDP per capita

```python
import sqlite3
conn = sqlite3.connect('country_statistics.db')
q = 'SELECT country, gdp_per_capita FROM country_statistics ORDER BY gdp_per_capita DESC LIMIT 1'
df_filtered_gdp_Per_Capita = pd.read_sql(q, conn)
conn.close()
print(df_filtered_gdp_Per_Capita)
print("\n")
```

#17 Export the final merged dataset as a CSV file

```python
import sqlite3
conn = sqlite3.connect('country_statistics.db')
final_df = pd.read_sql_query("SELECT * FROM country_statistics", conn)
conn.close()
final_df.to_csv('country_statics.csv', index = False)
print("Dataset exported to 'country_statistics.csv',")
print("\n")
```
