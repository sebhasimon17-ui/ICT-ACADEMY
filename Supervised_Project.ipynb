# IQR Method

```python
# handling outlier in sqft living using IQR Method

q1 = house_data['sqft_living'].quantile(0.25)
q3 = house_data['sqft_living'].quantile(0.75)
print(f'Q1 = {q1}\nq3 = {q3}')
```

```python
iqr = q3 - q1
up_lim = q3 +(1.5*iqr)
low_lim = q1 - (1.5*iqr)
print(f'IQR = {iqr}\nup_lim = {up_lim}\nlow_lim = {low_lim}')
```

```python
house_data.loc[house_data['sqft_living'] > up_lim, 'sqft_living'] = up_lim
house_data.loc[house_data['sqft_living'] < low_lim, 'sqft_living'] = low_lim
```

```python
house_data[house_data['sqft_living']> up_lim].index
```

```python
outliers = house_data[(house_data['sqft_living']> up_lim) | (
    house_data['sqft_living']<low_lim
)]
```

```python
outliers
```

# Scalling and splitting

```python
x =house_data.drop('price', axis = 1)
y = house_data['price']
```

```python
x1 =x.copy()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x1 = scaler.fit_transform(x1)
```

```python
x1
```

```python
# Train test split
from sklearn.model_selection import train_test_split
x_train1, x_test1, y_train, y_test = train_test_split(x1, y, test_size = 0.2, random_state = 42)
```

```python
# Non scaled x train and y train

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

# Regression
### Linear regression

```python
# Linear Regression
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(x_train1, y_train)
y_pred_lr_model = lr_model.predict(x_test1)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
print('r2_score = ' , r2_score(y_test, y_pred_lr_model))
print('mean_absolute_error = ' , mean_absolute_error(y_test, y_pred_lr_model))
print('mean_squared_error = ' , mean_squared_error(y_test, y_pred_lr_model))
```

# Lasso Regression

```python
from sklearn.linear_model import Lasso
lasso_reg = Lasso(alpha=10, max_iter=10000)
lasso_reg.fit(x_train1, y_train)
y_pred_lasso_model = lasso_reg.predict(x_test1)
print('r2_score:', r2_score(y_test, y_pred_lasso_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_lasso_model))
print('mean_squared_error:', mean_squared_error(y_test, y_pred_lasso_model))
```

# SVM

```python
from sklearn.svm import SVR
svr_model = SVR()
svr_model.fit(x_train1, y_train)
y_pred_svr_model = svr_model.predict(x_test1)
print('r2_score:', r2_score(y_test, y_pred_svr_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_svr_model))
print('mean_squared_error:', mean_squared_error(y_test, y_pred_svr_model))
```

```python
# Polynominal Kernal

from sklearn.svm import SVR
svr_model = SVR(kernel='poly')
svr_model.fit(x_train1, y_train)
y_pred_svr_model = svr_model.predict(x_test1)
print('r2_score:', r2_score(y_test, y_pred_svr_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_svr_model))
print('mean_squared_error:', mean_squared_error(y_test, y_pred_svr_model))
```

```python
# RBF Kernal

svr_model = SVR(kernel='rbf')
svr_model.fit(x_train1, y_train)
y_pred_svr_model = svr_model.predict(x_test1)
print('mean_squared_error:', mean_squared_error(y_test, y_pred_svr_model))
print('r2_score:', r2_score(y_test, y_pred_svr_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_svr_model))
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
dt_model = DecisionTreeRegressor()
dt_model.fit(x_train1, y_train)
y_pred_dt_model = dt_model.predict(x_test1)
print('r2_score:', r2_score(y_test, y_pred_dt_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_dt_model))
print('mean_squared_error:', mean_squared_error(y_test, y_pred_dt_model))
```

# Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor()
rf_model.fit(x_train1, y_train)
y_pred_rf_model = rf_model.predict(x_test1)
print('r2_score:', r2_score(y_test, y_pred_rf_model))
print('mean_absolute_error:', mean_absolute_error(y_test, y_pred_rf_model))
print('mean_squared_error:', mean_squared_error(y_test, y_pred_rf_model))
```
