```python
# handling outliers in sqft_living
```

```python
# Outlier handling using IQR method

q1 = house_df['sqft_living'].quantile(0.25)
q2 = house_df['sqft_living'].quantile(0.50)
q3 = house_df['sqft_living'].quantile(0.75)

print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')

iqr = q3 - q1

up_lim = q3 + (1.5 * iqr)
low_lim = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

```python
house_df.loc[house_df['sqft_living'] > up_lim,'sqft_living'] = up_lim

house_df.loc[house_df['sqft_living'] < low_lim,'sqft_living'] = low_lim
```

# KMeans Clustering

```python
#  using elbow method to find the optimum number of clusters
from sklearn.cluster import KMeans

wcss = []
for i in range(3,9):
  kmeans =KMeans(n_clusters=i, init= 'k-means++', random_state=42)
  kmeans.fit(x)
  wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(3,9),wcss,'o-')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS value')
plt.grid()
```

```python
# elbow is appearing at k=4
# we can create kmeans model with no. of clusters=4

kmeans1 = KMeans(n_clusters=4, init='k-means++', random_state=42)
y_kmeans = kmeans1.fit_predict(x)
```

```python
print(y_kmeans)
```

```python
#silhoutte score for kmean clustering

from sklearn.metrics import silhouette_score
sil_kmeans =  silhouette_score(x,y_kmeans)
print(sil_kmeans)
```

# Agglomerative hierarchichal clustering

```python
# using dentrogram to find the optimum number of clusters
import scipy.cluster.hierarchy as sch

dentrogram = sch.dendrogram(sch.linkage(x,method='ward'))
plt.title('Dendrogram')
plt.xlabel('Customer')
plt.ylabel('Euclidean Distance')
plt.show()
```

# PCA

```python
#applying scaling

scalar = StandardScaler()
scaled_data = scalar.fit_transform(x1)
# values are stored as numpy array
scaled_data = pd.DataFrame(scaled_data,columns = x1.columns)
#converted to Dataframe
scaled_data.describe()
```

```python
from sklearn.decomposition import PCA
pca = PCA().fit(scaled_data)      #pca
```

```python
plt.rcParams["figure.figsize"] = (12,6)
fig, ax= plt.subplots()
xi = np.arange(1,12,step=1)
yi = np.cumsum(pca.explained_variance_ratio_)
plt.ylim(0.0,1.1)
plt.plot(xi,yi,marker = 'o', linestyle = '--',color ='b')
plt.xlabel('Number of components')
plt.xticks(np.arange(0,12,step=1))
plt.ylabel('Cumulative variance(%)')
plt.title('The number of components needed to explain variance')
plt.axhline(y=0.97, color ='r',linestyle = '-')
plt.text(0.5,0.85,'97% cut-off threshold', color ='red', fontsize =16)
ax.grid(axis ='x')
plt.show()
```

```python
pca = PCA(n_components=9)     # reduce dimensions to 9
pca.fit(scaled_data)

# Transform the dataset
X_pca = pca.transform(scaled_data)

pca.get_covariance()    # covarience matrics
explained_variance = pca.explained_variance_ratio_
explained_variance
```

```python
df_pca =pd.DataFrame(X_pca,columns = ['PC1','PC2','PC3','PC4','PC5','PC6','PC7','PC8','PC9'])
df_pca
```

```python
# KMeans
# elbow is appearing at k=4
# we can create kmeans model with no. of clusters=4

kmeans1 = KMeans(n_clusters=4, init='k-means++', random_state=42)
y_kmeans = kmeans1.fit_predict(df_pca)
```

```python
#silhoutte score for kmean clustering
from sklearn.metrics import silhouette_score
sil_kmeans =  silhouette_score(df_pca,y_kmeans)
print(sil_kmeans)
```

```python
# Agglomerative hierarchichal clustering
dendrogram = sch.dendrogram(sch.linkage(df_pca, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distance')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
hc = AgglomerativeClustering(n_clusters=2, linkage = 'ward')
y_hc = hc.fit_predict(df_pca)
```
